## Data Cleaning

```python
term_60 = (df['term'].str.strip()== '60 months' )
df['term_60']= term_60.astype('int8')
```

```python
emp_length_map = {
    '< 1 year': 0, '1 year': 1, '2 years': 2, '3 years': 3, '4 years': 4,
    '5 years': 5, '6 years': 6, '7 years': 7, '8 years': 8, '9 years': 9,
    '10+ years': 10
}
df['emp_length'] = df['emp_length'].map(emp_length_map).astype('Int8')
```

```python
df['loan_status'].value_counts()
```

```python
charged_off = df['loan_status'].str.strip() == 'Charged Off'
df['loan_default']= charged_off.astype('int8')
```

```python
drop_cols = ['term','loan_status','title','address']
df.drop(columns = drop_cols, inplace = True)
```
